# 5. Train / Test Split Data

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,
    random_state=42,
    stratify=y # Keep approximately the same class distribution in both training and testing sets.
)
```

```python
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
y_train.value_counts(normalize=True) * 100 # verify strarify
```

# 6. Preprocessing

### catagorical and numerical col

```python
categorical_cols = x_train.select_dtypes(include='object').columns
categorical_cols
```

```python
numerical_cols = x_train.select_dtypes(include=np.number).columns
numerical_cols
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

### OneHotEncoder -> Converts categorical values into numerical columns
### StandardScaler -> Standardizes numerical features so they have approximately
### ColumnTransformer -> Allows us to apply different preprocessing to different columns:

```python
num_transformer = StandardScaler()
```

```python
cat_transformer = OneHotEncoder(handle_unknown='ignore')
```

```python
preprocessor = ColumnTransformer(
    [
        ('num', num_transformer, numerical_cols),
        ('cat', cat_transformer, categorical_cols)
    ]
)
```

```python
preprocessor.fit(x_train)
```

```python
x_train_transformed = preprocessor.transform(x_train)
```

```python
x_test_transformed = preprocessor.transform(x_test)
```

# Model 1. LogisticRegression

```python
from sklearn.linear_model import LogisticRegression
```

```python
log_model = LogisticRegression(max_iter=1000)
# Create a Logistic Regression binary classification model and allow its optimization process up to 1000 iterations to find the model parameters.
```

**fit() means:**

Train the model using the provided input data and known answers.

**Why do we use x_train_transformed?**

Because we already converted:

numerical features → scaled values
categorical features → one-hot encoded values

**Why y_train?**
Because this is supervised learning. The model needs the actual answers:

```python
log_model.fit(x_train_transformed, y_train)
```

**Why do we use x_test_transformed?**

Because these are customers the model didn't train on.

That's important because we want to know how well the model performs on unseen data.

```python
y_pred = log_model.predict(x_test_transformed)
```

**Get purchase probabilities**

Before evaluating the predicted classes, let's get something more useful for our hackathon: the probability of purchase.

```python
y_prob = log_model.predict_proba(x_test_transformed)[:, 1]
```

```python
y_prob
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

**Why these metrics?**
We'll compare the model's y_pred against the actual y_test:

Accuracy → overall percentage of correct predictions

Precision → among predicted purchasers, how many actually purchased

Recall → among actual purchasers, how many we successfully detected

F1-score → balance between precision and recall

```python
accuracy_score(y_test, y_pred)
# accuracy_score() compares the actual answers (y_test) with the model's predictions (y_pred).
# Accuracy=Correct Predictions / Total Predictions
```

```python
precision_score(y_test, y_pred)
# What it measures: Among all customers the model predicted as Purchase = 1, what percentage actually purchased?
```

```python
recall_score(y_test, y_pred)
# Recall measures how many of the customers who actually purchased were correctly identified by the model.
```

```python
f1_score(y_test, y_pred)
# F1-score combines precision and recall into a single metric.
# f1 = 2 * [(precision * recall) / (precision + recall)]
```

Confusion Matrix

Now we need to see exactly what types of mistakes the model is making.

A confusion matrix shows exactly how our classification model's predictions compare with the actual outcomes

TN — True Negative: Customer didn't purchase → model correctly says No Purchase.

FP — False Positive: Customer didn't purchase → model incorrectly says Purchase.

FN — False Negative: Customer purchased → model incorrectly says No Purchase.

TP — True Positive: Customer purchased → model correctly says Purchase.

```python
from sklearn.metrics import confusion_matrix
```

```python
confusion_matrix(y_test, y_pred)
```

```python
from sklearn.metrics import roc_auc_score
```

```python
roc_auc_score(y_test, y_prob)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
```

```python
rf_model = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
rf_model.fit(x_train_transformed, y_train)
```

```python
rf_pred = rf_model.predict(x_test_transformed)
rf_prob = rf_model.predict_proba(x_test_transformed)[:, 1]
print('Accuracy :', accuracy_score(y_test, rf_pred))
print('Precision:', precision_score(y_test, rf_pred))
print('Recall   :', recall_score(y_test, rf_pred))
print('F1 Score :', f1_score(y_test, rf_pred))
print('ROC-AUC  :', roc_auc_score(y_test, rf_prob))
```

```python
print('Confusion Matrix:')
print(confusion_matrix(y_test, rf_pred))
print('\nClassification Report:')
print(classification_report(y_test, rf_pred, target_names=['No Purchase', 'Purchase']))
```
